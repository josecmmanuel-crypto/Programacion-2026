# Problema #1: Calculadora de propina inteligente

## Problema

Escriba aquí la descripción del problema.

## Entrada

- Solicitar el total de la cuenta

## Proceso

-Comparar los numeros de la cuenta y sacar la propina, si es mayor la cuenta de 100.000 una propina de 15%, si es es mayor de 50.000 una propina de 10% y si es una cuenta menor de 49.999 una propina del 5%

## Salida esperada

-Se espera entregar la cuenta total y el calculo de la propina y el total sumando la propina

```python
cuenta = float(input("Ingrese el monto de la cuenta: "))
if cuenta >= 100000:
    propina = cuenta * 0.15
    print(f"La propina es de: {propina}")
    total = cuenta + propina
    print(f"El total a pagar es de: {total}")
elif cuenta >= 50000:
  propina = cuenta * 0.10
  print(f"La propina es de: {propina}")
  total = cuenta + propina
  print(f"El total a pagar es de: {total}")
else:
  propina = cuenta * 0.05
  print(f"La propina es de: {propina}")
total = cuenta + propina
print(f"El total a pagar es de: {total}")
```

# Problema #2: Detector de número par/impar

## Problema

El usuario me da un numero entero y necesita saber si es par o impar

## Entrada

-El usuario me da un numero entero

## Proceso

-El numero entero que me da el usuario tiene que ser divisible entre 2

## Salida esperada

-Se espera entregarle al cliente si el numero que dio es par o impar

```python
numero = int(input("Ingrese un numero entero: "))
if numero % 2 == 0:
    print("El numero es par")
else:
    print("El numero es impar")
```

# Problema #3: Clasificador de edad

## Problema

Se necesita clasificar a una persona en tres grupos dependiendo de su edad en niño, menor de edad, mayor de edad y adulto mayor

## Entrada

-Se necesita saber la edad del usuario

## Proceso

-Teniendo la edad del usuario y clasifica en los diferentes grupos

## Salida esperada

-Se espera entregar el cual grupo esta el usuario

```python
edad = int(input("Ingrese su edad: "))
if edad < 18:
    print("Usted es menor de edad")
elif edad >= 18 and edad < 65:
    print("Usted es adulto")
else:
    print("Usted es adulto mayor")
```

# Problema #4: Validador de compra con descuento

## Problema

Un restaurante quiere aplicar descuentos al cliente dependiendo si es cliente frecuente se le debe aplicar un descuento del 15%

## Entrada

-Se nesecita si el cliente es cliente frecuente y asi saber el monto que se le debe aplicar el descuento

## Proceso

-Se verifica si es cliente frecuente y si lo es se le debe aplicar el descuento

## Salida esperada

-Se le dara el monto y si se le aplicara el descuento

## Pruebas manuales


```python
Cuenta = float(input("Ingrese el monto de la cuenta: "))
Cliente = input("¿Es cliente frecuente? (si/no): ").strip().lower()
if Cliente == "si":
    Descuento = Cuenta * 0.15
    print(f"El descuento es de: {Descuento}")
    Total = Cuenta - Descuento
    print(f"El total a pagar es de: {Total}")
else:
    print("No hay descuento")
    Total = Cuenta
    print(f"El total a pagar es de: {Total}")
```
